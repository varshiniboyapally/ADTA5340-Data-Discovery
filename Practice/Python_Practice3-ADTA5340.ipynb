MERGING, JOINING, AND CONCATENATING

```python
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3'],
                        'C': ['C0', 'C1', 'C2', 'C3'],
                        'D': ['D0', 'D1', 'D2', 'D3']},
                        index=[0, 1, 2, 3])
df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                        'B': ['B4', 'B5', 'B6', 'B7'],
                        'C': ['C4', 'C5', 'C6', 'C7'],
                        'D': ['D4', 'D5', 'D6', 'D7']},
                         index=[4, 5, 6, 7]) 
df3 = pd.DataFrame({'A': ['A8', 'A9', 'A10', 'A11'],
                        'B': ['B8', 'B9', 'B10', 'B11'],
                        'C': ['C8', 'C9', 'C10', 'C11'],
                        'D': ['D8', 'D9', 'D10', 'D11']},
                        index=[8, 9, 10, 11])
```

```python
df1
```

```python
df2
```

```python
df3
```

```python
# to concatenate
# Note here that we are joining on the default axis = 0 ie stacking them horizontally
pd.concat([df1,df2,df3])
```

```python
# # stacking vertically:
pd.concat([df1,df2,df3],axis=1)
```

```python
# # stack vertically and fill all NaNs with zeros:
pd.concat([df1,df2,df3],axis=1).fillna(0)
```

MERGING DATA FRAMES - By default merging is done as 'inner' merge 

```python
left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
   
right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                          'C': ['C0', 'C1', 'C2', 'C3'],
                          'D': ['D0', 'D1', 'D2', 'D3']})  
```

```python
left
```

```python
right
```

```python
# this is equivalent to sql "inner join" where all common records from both dataFrames w.r.t "on" 
# are included in the results
pd.merge(left, right, how='inner', on='key')
```

```python
pd.merge(left, right, how='outer', on='key')
```

```python
pd.merge(left, right, how='cross')
```

```python
# more merging examples:
left = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                        'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3']})
    
right = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
                      'key2': ['K0', 'K0', 'K0', 'K0'],
                         'C': ['C0', 'C1', 'C2', 'C3'],
                         'D': ['D0', 'D1', 'D2', 'D3']})

my_left = pd.DataFrame({'prodid':['a','a','b','c'],
                        'customerid':['a','b','a','b'],
                        'num_sold':[4,20,12,10],
                        'unit_price':[20,15,40,16]})

my_right = pd.DataFrame({'prodid':['a','b','b','c'],
                        'customerid':['a','a','a','a'],
                        'num_manufacturers':[41,23,25,49],
                        'quantity_shipped':[47,52,16,30]})
```

```python
my_left
```

```python
my_right
```

```python
left
```

```python
right
```

```python
# bring all records from both data frames, where there is no match on "on", you assign data point of NaN
pd.merge(left, right, how = 'outer', on = ['key1','key2'])
```

```python
# that is, ONLY where the pair ['key1','key2'] match on both data frame is selected
pd.merge(left, right, how = 'inner', on = ['key1','key2'])
```

```python
# like sql right join
pd.merge(left, right, how = 'right', on = ['key1','key2'])
```

```python
# like sql left join
pd.merge(left, right, how = 'left', on = ['key1','key2'])
```

JOINING IN PANDAS

```python
# Note: JOin is done "on" an index instead of a column as in merge:
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                      index=['K0', 'K1', 'K2']) 

right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                    'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
```

```python
left
```

```python
right
```

```python
# will perform the join on "index" in the data frames
left.join(right)
```

```python
left.join(right, how = 'outer')
```

```python
left.join(right, how = 'right')
```
