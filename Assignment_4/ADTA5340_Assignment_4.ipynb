```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
# Specify the fields with their names as follows
fields = ['preg', 'plas', 'pres', 'skin', 'test', 'mass', 'pedi', 'age', 'class']
# Read the dataset (replace 'data.csv' with your file name)
df = pd.read_csv("Assignment_4pima_diabetes.csv", names= fields)
# Show first 5 rows to confirm
df.head()
```

```python
df.drop('class', axis=1,inplace=True)
```

```python
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_cols = df.select_dtypes(include='number').columns

# Loop through each numeric column and create a histogram
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col],color='skyblue', edgecolor='black')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
# Loop through each numeric column and create a Density plot
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    df[col].plot(kind='density', color='green')
    plt.title(f'Density Plot of {col}')
    plt.xlabel(col)
    plt.show()
```

```python
# Loop through each numeric column and create a Box plot
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col])
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)
    plt.show()
```

```python
from pandas.plotting import scatter_matrix

scatter_matrix(df[numeric_cols], alpha=0.5, figsize=(15,15))
plt.suptitle('Scatter Matrix of Numeric Variables')
plt.show()
```
