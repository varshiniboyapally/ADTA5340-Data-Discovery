### 'Avg. Area House Age': Avg Age of Houses in same city-suburb
### 'Avg. Area Number of Rooms': Avg Number of Rooms for Houses in same city-suburb
### 'Avg. Area Number of Bedrooms': Avg Number of Bedrooms for Houses in same city-suburb

LET US START MODELING!

```python
x = DentonHousing[['Avg. Area Income', 'Avg. Area House Age', 'Avg. Area Number of Rooms',
       'Avg. Area Number of Bedrooms', 'Area Population']]
y = DentonHousing['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
# splitting the data into training and testing datasets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.4, random_state=101)
```

## CREATING AND TRAINING THE LINEAR REGRESSION MODEL

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
```

```python
# we now train our model
lm.fit(x_train, y_train)
```

## MODEL EVALUATION

```python
# print the intercept
print(lm.intercept_)
```

```python
lm.coef_
```

```python
# x.columns
coeff_df = pd.DataFrame(lm.coef_, x.columns, columns=['Coefficient'])
```

```python
coeff_df
```

## PREDICTION FROM THE MODEL

```python
predictions = lm.predict(x_test)
```

```python
plt.scatter(y_test,predictions)
```

## RESIDUAL PLOT

```python
# you can also experiment with diff. number of bins and observe the effects
sns.displot((y_test-predictions),kind='hist',kde=True,bins=50)
#plt.show()
```

```python
from sklearn import metrics
```

```python
print('MAE:', metrics.mean_absolute_error(y_test, predictions))
print('MSE:', metrics.mean_squared_error(y_test, predictions))
print('RMSE:', np.sqrt(metrics.mean_squared_error(y_test, predictions)))
```
