```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", palette="muted", font_scale=1.1)

# Loading the Boston Housing dataset with relevant column 
df = pd.read_csv('housing_boston.csv', usecols=['rm', 'age', 'dis', 'rad', 'ptratio', 'medv'])

# displaying dataset
df.head()
```

```python
# showing Histogram of RM
sns.histplot(df['rm'], bins=20, kde=True)
plt.title("Distribution of rm")
#Display plot
plt.show()
```

```python
# showing Boxplot of RM
sns.boxplot(df['rm'])
plt.title("Boxplot of rm")
#Display plot
plt.show()
```

### The histogram shows that most homes have between 5 and 7 rooms, with a bell-shaped distribution.

### The boxplot highlights a few outliers above 8 rooms, suggesting a small number of very large, high-end homes.


```python
# Displaying Histogram of AGE
sns.histplot(df['age'], bins=20, kde=True)
plt.title("Distribution of age")
#Display plot
plt.show()
```

```python
# Displaying Boxplot of AGE
sns.boxplot(df['age'])
plt.title("Boxplot of age")
#Display plot
plt.show()
```

### The distribution is skewed toward high values, meaning many neighborhoods consist of older housing stock (over 80% pre-1940 units).

### The boxplot confirms a concentration of old houses, which might influence lower property values in some areas.

```python
# Displaying Histogram of DIS
sns.histplot(df['dis'], bins=20, kde=True)
plt.title("Distribution of dis")
#Display plot
plt.show()
```

```python
# Displaying Boxplot of DIS
sns.boxplot(df['dis'])
plt.title("Boxplot of dis")
#Display plot
plt.show()
```

### Most neighborhoods are located between 2 and 5 units of distance from employment centers, showing clustering in mid-distance suburbs.

### A few neighborhoods are much farther away, which could represent wealthier suburban communities or rural-like areas.

## 2. Perform bivariate visualizations between the median value of owner-occupied homes, medv and the following variables on the dataset: rm, age, dis, rad and ptratio. Please use different Jupyter Notebook cell for each bivariate visualization, and briefly comment on your results

```python
# Displaying scatterplot for rm vs medv
sns.scatterplot(x=df['rm'], y=df['medv'])
plt.title("rm vs medv")
plt.xlabel("Average Rooms (rm)")
plt.ylabel("Median Value (medv)")
#Display plot
plt.show()
```

### We can see a significant increasing pattern in home with more rooms being closely associated with high median values. This says that larger homes seem to be more attractive that indicates rich neighborhoods.


```python
# Displaying scatterplot for age vs medv
sns.scatterplot(x=df['age'], y=df['medv'])
plt.title("age vs medv")
plt.xlabel("Age of Units (%)")
plt.ylabel("Median Value (medv)")
#Display plot
plt.show()
```

### The scatterplot shows that old houses are likely to have lower values, however the relation isn’t precisely linear. This could be because of decreases in value, old infrastructure, and outdated interior designs that diminishing demand in the market.


```python
# Displaying scatterplot for dis vs medv
sns.scatterplot(x=df['dis'], y=df['medv'])
plt.title("dis vs medv")
plt.xlabel("Distance to Employment Centers (dis)")
plt.ylabel("Median Value (medv)")
# Display plot
plt.show()
```

### Homes which are located farther away from employment centers (higher DIS) tend to have higher median values. It additionally suggests that rural areas or less crowded locations could be wealthier, since families opt for space and commute into urban cities.


```python
# Displaying scatterplot for rad vs medv
sns.scatterplot(x=df['rad'], y=df['medv'])
plt.title("rad vs medv")
plt.xlabel("Highway Accessibility (rad)")
plt.ylabel("Median Value (medv)")
#Display plot
plt.show()
```

### The median home values vary drastically among various levels of highway accessibility, yet specific categories such as RAD=24 have lower values.This could suggest that though accessibility is essential, living close enough to a highway might reduce desirability due to various pollutions.


```python
# Displaying scatterplot for ptratio vs medv
sns.scatterplot(x=df['ptratio'], y=df['medv'])
plt.title("ptratio vs medv")
plt.xlabel("Pupil-Teacher Ratio")
plt.ylabel("Median Value (medv)")
#Display plot
plt.show()
```

### The scatterplot depicts a negative trend which shows higher pupil-teacher ratios (crowded schools) have a correlation with lower home values. This means that costs of housing are greatly impacted by school quality, with families spending higher to stay in neighborhoods with quality educational systems.
